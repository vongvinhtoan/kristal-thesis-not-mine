```python
(
    cleaned_price_df.sort("ticker", "date")
    .with_columns(
        pl.col("close").shift(1).over("ticker").alias("prev_close"),
    )
    .with_columns(
        (pl.col("close") / pl.col("prev_close")).log().alias("dret"),
    )
    .with_columns(
        pl.when(pl.col("mktcap") == 0)
        .then(None)
        .otherwise(pl.col("mktcap"))
        .alias("mktcap"),
    )
    .with_columns(
        pl.col("date").dt.year().alias("year"),
        pl.col("date").dt.week().alias("week"),
    )
    .group_by("ticker", "year", "week")
    .agg(
        pl.col("dret").sum(),
        pl.col("mktcap").last().fill_null(strategy="backward").last().alias("mktcap"),
        pl.col("icb1").last().fill_null(strategy="backward").last(),
        pl.col("icb2").last().fill_null(strategy="backward").last(),
        pl.col("icb3").last().fill_null(strategy="backward").last(),
        pl.col("icb4").last().fill_null(strategy="backward").last(),
        pl.col("icb5").last().fill_null(strategy="backward").last(),
    )
)
```

```python
cleaned_price_df = (
    price_df.join(
        icb_df,
        on="ticker",
        how="left",
        validate="m:1",
    )
    .filter(pl.col("icb2").is_not_null())
    .filter(
        ~pl.col("icb2").is_in(
            [
                "Dịch vụ tài chính",
                "Ngân hàng",
                "Bảo hiểm",
            ]
        )
    )
    .filter(~pl.col("ticker").is_in(["DSC", "MIC"]))
    .with_columns(
        pl.when(pl.col("mktcap") != 0)
        .then(pl.col("mktcap"))
        .otherwise(None)
        .alias("mktcap"),
        year=pl.col("date").dt.year(),
        week=pl.col("date").dt.week(),
        dayofweek=pl.col("date").dt.weekday(),
    )
    .filter(~pl.col("dayofweek").is_in([6, 7]))
    .filter(pl.col("year") >= 2010)
    .filter(pl.col("year") <= 2024)
)
cleaned_price_df
```

```python
wret_df = (
    cleaned_price_df.sort(["ticker", "date"])
    .upsample(time_column="date", every="1d", group_by="ticker", maintain_order=True)
    .with_columns(
        dret=(pl.col("close") / pl.col("close").shift(1)).log().over("ticker")
    )
    .filter(pl.col("close").is_not_null() | pl.col("dret").is_not_null())
    .group_by(["ticker", "year", "week"], maintain_order=True)
    .agg(
        [
            pl.col("dret").sum(),
            pl.col("mktcap").last(ignore_nulls=True),
            pl.col("icb1").last(ignore_nulls=True),
            pl.col("icb2").last(ignore_nulls=True),
            pl.col("icb3").last(ignore_nulls=True),
            pl.col("icb4").last(ignore_nulls=True),
            pl.col("icb5").last(ignore_nulls=True),
        ]
    )
    # 6. Calculate weekly return and filter outliers
    .with_columns(wret=(pl.col("dret").exp() - 1))
    .filter(pl.col("wret").abs() < (pow(1.1, 5) - 1))
)
wret_df
```

```python
vwmktret_df = (
    wret_df
    # Value-weighted market return
    .sort("ticker", "year", "week")
    .with_columns(
        pl.col("mktcap").sum().over("year", "week").alias("totalmktcap"),
    )
    .with_columns(
        (pl.col("mktcap") / pl.col("totalmktcap")).alias("mktcap_weighted"),
    )
    .with_columns(
        pl.int_range(1, pl.len() + 1).over("ticker").alias("weekid"),
    )
    .with_columns(
        pl.col("mktcap_weighted").shift(1).over("ticker").alias("mktcap_weighted_L1"),
    )
    .with_columns(
        (pl.col("mktcap_weighted_L1") * pl.col("wret")).alias("wret_weighted"),
    )
    .with_columns(
        pl.col("wret_weighted").sum().over("year", "week").alias("vwmktret"),
    )
    # Value-weighted industry return
    # .with_columns(
    #     pl.col("mktcap").sum().over(["icb1", "year", "week"]).alias("total_mktcap_ind")
    # )
    # .with_columns((pl.col("mktcap") / pl.col("total_mktcap_ind")).alias("ind_weight"))
    # .with_columns(pl.col("ind_weight").shift(1).over("ticker").alias("ind_weight_L1"))
    # .with_columns((pl.col("ind_weight_L1") * pl.col("wret")).alias("ind_wret_weighted"))
    # .with_columns(
    #     pl.col("ind_wret_weighted")
    #     .sum()
    #     .over(["icb1", "year", "week"])
    #     .alias("vwindret")
    # )
    # .with_columns(
    #     [
    #         pl.col("vwmktret").shift(1).over("ticker").alias("vwmktret_L1"),
    #         pl.col("vwmktret").shift(-1).over("ticker").alias("vwmktret_F1"),
    #         pl.col("vwindret").shift(1).over("ticker").alias("vwindret_L1"),
    #         pl.col("vwindret").shift(-1).over("ticker").alias("vwindret_F1"),
    #     ]
    # )
)
vwmktret_df
```
