```python
df1 = pd.read_stata(DATA_DIR/"temp/1-100_hnx_2007-2010.xlsx.dta")
df1 = pl.from_pandas(df1)
df1 = df1.with_columns(pl.col("date").cast(pl.Date))
df1 = df1.with_columns(pl.col("volume").cast(pl.Int64))
df1 = df1.with_columns(pl.col("shares").cast(pl.Int64))
df1 = df1.with_columns(pl.col("mktcap").cast(pl.Int64))
df1
```

```python
df2 = pl.read_parquet(DATA_DIR.parent / "processed/cleanup/1-100_HNX_2007-2010.parquet")
df2
```

```python
df1.join(df2, on=df1.columns, how="anti")
```

```python
df2.join(df1, on=df2.columns, how="anti")
```

```python
left_only = df1.join(df2, on=df1.columns, how="anti")
right_only = df2.join(df1, on=df1.columns, how="anti")
diff = pl.concat([left_only, right_only])
diff
```

```python
fill = {"close": -1.0, "volume": -1, "shares": -1, "mktcap": -1}

left = df1.with_columns([pl.col(c).fill_null(v) for c, v in fill.items()])
right = df2.with_columns([pl.col(c).fill_null(v) for c, v in fill.items()])

left_only = left.join(right, on=left.columns, how="anti")
right_only = right.join(left, on=right.columns, how="anti")
diff = pl.concat([left_only, right_only])

diff
```
