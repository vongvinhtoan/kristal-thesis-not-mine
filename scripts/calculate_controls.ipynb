```python
import polars as pl
from pathlib import Path
```

```python
SCRIPT_DIR = Path(".").absolute().resolve()
DATA_DIR = (SCRIPT_DIR.parent / "data").absolute().resolve()
PROCESSED_DATA_DIR = (SCRIPT_DIR.parent / "processed").absolute().resolve()

print(f"{DATA_DIR=}")
print(f"{PROCESSED_DATA_DIR=}")
assert DATA_DIR.is_dir()
assert PROCESSED_DATA_DIR.is_dir()
```

```python
financials_df = pl.read_parquet(PROCESSED_DATA_DIR / "financials.parquet")
financials_df
```

```python
icb_df = pl.read_parquet(PROCESSED_DATA_DIR / "icb.parquet")
icb_df
```

```python
cleaned_financials_df = (
    financials_df.join(
        icb_df,
        on="ticker",
        how="left",
        validate="m:1",
    )
    .filter(pl.col("icb2").is_not_null())
    .filter(
        ~pl.col("icb2").is_in(
            [
                "Dịch vụ tài chính",
                "Ngân hàng",
                "Bảo hiểm",
            ]
        )
    )
    .filter(~pl.col("ticker").is_in(["DSC", "MIC"]))
    .sort(["ticker", "year"])
    .with_columns(
        [
            pl.col("TA").log().alias("SIZE"),
            (pl.col("mktcap") / pl.col("equity")).alias("MTB"),
            (pl.col("TL") / pl.col("TA")).alias("LEVRG"),
            (pl.col("netprofit") / pl.col("TA")).alias("ROA"),
        ]
    )
)
cleaned_financials_df
```
